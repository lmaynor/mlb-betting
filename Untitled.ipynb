```python
import requests, json

url = "https://api.actionnetwork.com/web/v1/scoreboard/mlb?period=game&bookIds=15,30,76,123&date=20260507"
r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=15)
print(r.status_code)
if r.status_code == 200:
    d = r.json()
    games = d.get("games", [])
    print(f"Games: {len(games)}")
    if games:
        print(json.dumps(games[0], indent=2)[:2000])
```

```python
import requests, json

url = "https://api.actionnetwork.com/web/v1/scoreboard/mlb?period=game&bookIds=15,30,76,123&date=20260507"
r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=15)
d = r.json()

game = d["games"][0]
print("Top-level keys:", list(game.keys()))
print("\nOdds sample:")
print(json.dumps(game.get("odds", [])[:3], indent=2))
```

```python
import requests, json

# DraftKings book_id on ActionNetwork is 3
url = "https://api.actionnetwork.com/web/v1/scoreboard/mlb?period=1&bookIds=3&date=20260507"
r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=15)
d = r.json()
games = d.get("games", [])
print(f"Games: {len(games)}")
for g in games[:3]:
    odds = g.get("odds", [])
    teams = [t["abbr"] for t in g.get("teams", [])]
    print(f"\n{teams}: {len(odds)} odds entries")
    if odds:
        print(json.dumps(odds[0], indent=2)[:500])
        
```
